```python
from omniuq import SpectralUncertainty, load_llm_model, load_openai_client
import os
from dotenv import load_dotenv
from omniuq import load_openai_client

load_dotenv()   # reads .env, populates os.environ

key = os.environ["OPENAI_API_KEY"]

tokenizer, model = load_llm_model("meta-llama/Llama-3.1-8B-Instruct")
clarifier = load_openai_client(api_key=key, model="gpt-4o")
judge = load_openai_client(api_key=key, model="gpt-4.1")

# uq = SpectralUncertainty(
#     tokenizer, model,
#     clarifier=clarifier,
#     judge=judge,
# )

uq = SpectralUncertainty(
    tokenizer, model,
    clarifier=(tokenizer, model),    # use the HF model itself
    judge=(tokenizer, model), max_clarifications=1, m_samples=1
    )
```

```python
uq.score("what is the capital of france?", verbose=True)
```
